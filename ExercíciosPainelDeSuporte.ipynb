```python
temperatura = int(input('Insira a temperatura: '))
umidade = int(input('Insira a umidade do ar, de 0% a 100%: '))
pressao = int(input('Insira o nível de pressão: \n [1] - ALTA \n [2] - ESTÁVEL \n [3] - BAIXA'))

if pressao == 3:
  print('Descompressão imediata!')
  if temperatura < 0:
    print('ALERTA: Congelamento a Vácuo Iminente!')
  else:
    print('ALERTA: Evacuar Cúpula por Perda de Pressão!')

elif pressao == 2:
  print('PRESSÃO ESTÁVEL: analisando temperatura interna ...')
  if (temperatura >= 18 and temperatura <= 26) and (umidade >= 40 and umidade <= 70):
      print('Ambiente Ideal: Bioma em Perfeito Equilíbrio.')
  else:
    print('Aviso: Clima Oscilante. Ativar Suporte de Carga.')

elif pressao == 1:
  print('Pressão alta detectada! ...')
  if temperatura > 40 or umidade > 80:
    print('PERIGO: Risco de Explosão Hidrotérmica!')
  else:
    print('Atenção: Pressão Elevada. Aliviando Válvulas.')

else:
  print('Erro Crítico: Sensor de Pressão Inoperante!')
```
